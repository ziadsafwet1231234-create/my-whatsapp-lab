```python
import streamlit as st

# --- واجهة البرنامج ---
st.set_page_config(page_title="Cyber Mentor AI", page_icon="🛡️")

st.title("🛡️ أكاديمية الذكاء الاصطناعي للأمن السيبراني")
st.write("مرحباً بك يا بطل! أنا مدربك الخاص، سأعلمك كيف تفكر كهاكر وتدافع كخبير.")

# --- القائمة الجانبية (خريطة الطريق) ---
menu = ["الرئيسية", "المستوى 1: أساسيات", "المستوى 2: الثغرات", "اسأل المدرب الذكي"]
choice = st.sidebar.selectbox("اختر مرحلة التعلم:", menu)

if choice == "الرئيسية":
    st.subheader("ماذا ستتعلم اليوم؟")
    st.info("اختر مستوى من القائمة الجانبية لتبدأ رحلتك.")

elif choice == "المستوى 1: أساسيات":
    st.header("🕵️ مرحلة جمع المعلومات (Reconnaissance)")
    st.write("أول خطوة في الهاكينج هي معرفة 'الأبواب المفتوحة' في الجهاز.")
    
    st.video("https://www.youtube.com/watch?v=isZX_XKpAnc") # مثال لفيديو تعليمي
    
    st.markdown("---")
    answer = st.text_input("تحدي: ما هو الأمر الذي نستخدمه في Linux لمعرفة الـ IP الخاص بنا؟")
    if answer == "ifconfig" or answer == "ip a":
        st.success("إجابة صحيحة! أنت الآن جاهز للمستوى التالي.")
    elif answer:
        st.error("حاول مرة أخرى.. فكر في أوامر الشبكة في Linux.")

elif choice == "اسأل المدرب الذكي":
    st.header("🤖 اسألني أي سؤال في الأمن السيبراني")
    user_q = st.text_input("اكتب سؤالك هنا (مثلاً: ما الفرق بين الهكر الأخلاقي والغير أخلاقي؟)")
    if user_q:
        # هنا سنضع كود الـ AI لاحقاً
        st.write("جاري تحليل سؤالك... (هنا سيقوم الذكاء الاصطناعي بالرد عليك)")
```
